# Day 5: Data Cleaning
**AI/ML Internship – Task 1**

This notebook cleans `student_data.csv`: handling missing values, removing duplicate rows, fixing inconsistent text labels, and renaming columns. The cleaned file is saved for use in later EDA and modeling tasks.

## 1. Load the raw dataset

```python
import pandas as pd

df = pd.read_csv("../data/student_data.csv")
print("Shape before cleaning:", df.shape)
df.head()
```

## 2. Handle missing values

First, see exactly how much is missing and where, then decide how to handle it.
For numeric score/attendance columns, filling with the **column median** is safer than the mean,
since it's less affected by outliers.

```python
df.isnull().sum()
```

```python
numeric_cols = ["Study Hours", "Attendance (%)", "Math Score", "Science Score", "English Score"]

for col in numeric_cols:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)
    print(f"Filled missing values in '{col}' with median = {median_value}")
```

```python
# Confirm no missing values remain
df.isnull().sum()
```

## 3. Remove duplicate rows

```python
print("Duplicate rows found:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

## 4. Fix inconsistent text values

The `Gender` column has inconsistent labels like `male`, `Male`, `female `, `Female` (with stray
capitalization and whitespace). Standardize these before any grouping or analysis.

```python
df["Gender"].unique()
```

```python
df["Gender"] = df["Gender"].str.strip().str.capitalize()
df["Gender"].unique()
```

## 5. Rename columns for consistency
Use lowercase, underscore-separated names, which are easier to work with in code.

```python
df = df.rename(columns={
    "Student ID": "student_id",
    "Name": "name",
    "Gender": "gender",
    "Study Hours": "study_hours",
    "Attendance (%)": "attendance_pct",
    "Math Score": "math_score",
    "Science Score": "science_score",
    "English Score": "english_score",
})
df.columns.tolist()
```

## 6. Final check

```python
print("Final shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nGender values:", df["gender"].unique())
df.head()
```

## 7. Save the cleaned dataset

```python
df.to_csv("../data/student_data_cleaned.csv", index=False)
print("Saved cleaned dataset to ../data/student_data_cleaned.csv")
```
