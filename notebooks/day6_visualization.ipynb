# Day 6 – Data Visualization with Matplotlib & Seaborn

**Goal:** create at least 8 different charts, each with a title and axis labels, and write an observation after each one.

Dataset: Iris (loaded via Seaborn). Swap `df` for your `student_data.csv` if you want to reuse these on the mini project.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
os.makedirs('../images', exist_ok=True)

df = sns.load_dataset('iris')
df.head()
```

## 1. Histogram

```python
plt.figure(figsize=(7,4))
plt.hist(df['sepal_length'], bins=15, color='steelblue', edgecolor='black')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.savefig('../images/01_histogram.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _write what you see (shape, range, most common values)._

## 2. KDE / Histogram by species

```python
plt.figure(figsize=(7,4))
sns.histplot(data=df, x='petal_length', hue='species', kde=True)
plt.title('Petal Length Distribution by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Count')
plt.savefig('../images/02_hist_kde_species.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _which species are clearly separated?_

## 3. Scatter plot

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.savefig('../images/03_scatter.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _..._

## 4. Box plot (outliers)

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x='species', y='sepal_width')
plt.title('Sepal Width by Species')
plt.xlabel('Species')
plt.ylabel('Sepal Width (cm)')
plt.savefig('../images/04_boxplot.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _any outliers? which species?_

## 5. Violin plot

```python
plt.figure(figsize=(7,5))
sns.violinplot(data=df, x='species', y='petal_length')
plt.title('Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.savefig('../images/05_violin.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _..._

## 6. Bar chart (average per group)

```python
means = df.groupby('species')['petal_width'].mean()
plt.figure(figsize=(7,4))
means.plot(kind='bar', color=['#4c72b0','#55a868','#c44e52'], edgecolor='black')
plt.title('Average Petal Width by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Width (cm)')
plt.xticks(rotation=0)
plt.savefig('../images/06_bar.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _..._

## 7. Correlation heatmap

```python
corr = df.drop(columns='species').corr()
plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Iris Features')
plt.savefig('../images/07_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _which features are strongly correlated?_

## 8. Pair plot

```python
g = sns.pairplot(df, hue='species')
g.figure.suptitle('Pair Plot of Iris Features', y=1.02)
g.savefig('../images/08_pairplot.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Observation:** _..._

## 9. (Bonus) Count plot

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='species')
plt.title('Number of Samples per Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.savefig('../images/09_countplot.png', dpi=150, bbox_inches='tight')
plt.show()
```
