## 7. Practice Exercises (solved)
1. Display the shape of the dataset (rows, columns).
2. Show only the `Name` and `Math Score` columns for the first 5 rows.
3. Find the student(s) with the highest Science Score.
4. Count how many students have Attendance below 60%.
5. Find the average Study Hours grouped by Gender.

```python
# Exercise 1: shape of the dataset
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Exercise 2: Name and Math Score, first 5 rows
df[["Name", "Math Score"]].head(5)
```

```python
# Exercise 3: student(s) with the highest Science Score
top_science = df[df["Science Score"] == df["Science Score"].max()]
top_science[["Name", "Science Score"]]
```

```python
# Exercise 4: students with Attendance below 60%
low_attendance = df[df["Attendance (%)"] < 60]
print("Number of students with attendance below 60%:", len(low_attendance))
low_attendance[["Name", "Attendance (%)"]]
```

```python
# Exercise 5: average Study Hours grouped by Gender
df.groupby("Gender")["Study Hours"].mean()
```

## 8. Observations

- **Math** has the highest average score among the three subjects, followed by Science and then English (see section 5).
- `Gender` contains inconsistent labels (`Male`, `male`, `Female`, `female `) that must be standardized in Day 5 before grouping by gender is reliable.
- Several columns (`Study Hours`, `Attendance (%)`, `Math Score`, `Science Score`, `English Score`) have missing values, confirmed in section 6.
- The dataset contains duplicate rows, also confirmed in section 6, which will be removed during cleaning.
- Attendance below 60% is relatively rare, suggesting most students attend regularly, though this should be checked against score trends later.
