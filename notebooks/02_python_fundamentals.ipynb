## 1. Variables and Data Types
A variable is a name that stores a value. Python figures out the type automatically.

```python
name = "Krish"          # str  (text)
age = 21                # int  (whole number)
height = 1.75           # float (decimal number)
is_student = True       # bool (True / False)

print(name, age, height, is_student)
print(type(name), type(age), type(height), type(is_student))
```

```python
# Type conversion
score_text = "85"
score = int(score_text)       # str -> int
print(score + 10)             # 95

print(float(age))             # 21.0
print(str(height) + " m")     # 1.75 m
```

## 2. Operators

```python
a, b = 17, 5

print("Add:", a + b)
print("Subtract:", a - b)
print("Multiply:", a * b)
print("Divide:", a / b)          # 3.4
print("Floor divide:", a // b)   # 3
print("Remainder:", a % b)       # 2
print("Power:", a ** 2)          # 289
```

```python
# Comparison and logical operators
marks = 72
print(marks > 50)                 # True
print(marks == 100)               # False
print(marks >= 40 and marks < 80) # True
print(not (marks > 90))           # True
```

## 3. Conditional Statements

```python
def grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"

for m in [95, 80, 65, 45, 20]:
    print(m, "->", grade(m))
```

## 4. Loops

```python
# for loop over a range
for i in range(1, 6):
    print("Square of", i, "is", i ** 2)
```

```python
# while loop
count = 3
while count > 0:
    print("Countdown:", count)
    count -= 1
print("Go!")
```

```python
# break and continue
for n in range(1, 11):
    if n % 2 == 0:
        continue      # skip even numbers
    if n > 7:
        break         # stop the loop
    print(n)
```

## 5. Functions
Functions package reusable logic. Good functions do one job and have a clear name.

```python
def average(numbers):
    """Return the average of a list of numbers."""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

print(average([70, 85, 90, 65]))
```

```python
def describe_student(name, marks, passed_mark=40):
    status = "Pass" if marks >= passed_mark else "Fail"
    return f"{name} scored {marks} ({status})"

print(describe_student("Asha", 78))
print(describe_student("Ravi", 35))
print(describe_student("Meera", 55, passed_mark=60))
```

## 6. Lists
Ordered, changeable collections. Lists are everywhere in data work.

```python
marks = [72, 85, 60, 91, 55]

print(marks[0])        # first item
print(marks[-1])       # last item
print(marks[1:4])      # slice: items 1 to 3

marks.append(88)       # add to end
marks.remove(60)       # remove a value
marks.sort()           # sort in place
print(marks)
print("Highest:", max(marks), "Lowest:", min(marks), "Count:", len(marks))
```

```python
# List comprehension: build a new list in one line
marks = [72, 85, 60, 91, 55]
passed = [m for m in marks if m >= 60]
scaled = [m / 100 for m in marks]
print(passed)
print(scaled)
```

## 7. Tuples
Like lists, but **immutable** (cannot be changed). Good for fixed values such as coordinates.

```python
point = (10, 20)
print(point[0], point[1])

x, y = point            # unpacking
print("x =", x, "y =", y)

# point[0] = 5   # uncommenting this raises a TypeError
```

## 8. Dictionaries
Key-value pairs. Very similar to how a row of a dataset looks.

```python
student = {
    "name": "Asha",
    "age": 20,
    "marks": {"math": 88, "science": 92, "english": 79}
}

print(student["name"])
print(student["marks"]["math"])

student["city"] = "Jaipur"          # add a new key
print(student.get("grade", "N/A"))  # safe lookup with default

for subject, score in student["marks"].items():
    print(subject, "->", score)
```

## 9. Sets
Unordered collections of **unique** items. Handy for removing duplicates.

```python
cities = ["Jaipur", "Delhi", "Jaipur", "Mumbai", "Delhi"]
unique_cities = set(cities)
print(unique_cities)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
```

## 10. File Handling
Read and write text files. The `with` block closes the file automatically.

```python
import os

os.makedirs("../data", exist_ok=True)
path = "../data/day2_demo.txt"

# Write
with open(path, "w") as f:
    f.write("Name,Marks\n")
    f.write("Asha,88\n")
    f.write("Ravi,72\n")

# Read the whole file
with open(path, "r") as f:
    print(f.read())
```

```python
# Read line by line and process
total = 0
count = 0
with open(path, "r") as f:
    next(f)                       # skip the header line
    for line in f:
        name, mark = line.strip().split(",")
        total += int(mark)
        count += 1

print("Average marks:", total / count)
```
